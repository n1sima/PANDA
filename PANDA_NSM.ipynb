The energy transfer is

$$\Delta E = \frac{\hbar^2}{2m_n}\left(k_i^2 - k_f^2\right),$$

where $m_n$ is the neutron mass and $k_i$, $k_f$ are the incident and scattered
neutron wavenumbers. In the instrument $\frac{\hbar^2}{2m_n}$ is replaced by $ 2.072 \ \mathbf{meV}$ $\mathbf{fm}^2$ The scattering vector (momentum transfer) is

$$\vec{Q} = \vec{k}_i - \vec{k}_f.$$

The code determines the two spectrometer angles — the sample scattering angle `stt`
(2θ) and the sample rotation `sth` (θ) — automatically, but only if the values of
$k_i$, $k_f$ and $\vec{Q}$ are supplied. Otherwise, `stt` and `sth` must be given
explicitly.

The phonon_simple component should display a sample for phonon scattering in a fcc crystal, the dispersion condition is calculated individually for each neutron

## Phonon measurement in Panda 

The key idea is to measure the detector intensity at a fixed wavevector,either
the final wavevector kf = 2.66 Å⁻¹, or, in the other case, the initial wavevector
ki = 2.66 Å⁻¹, for a range of energy transfers dE. When kf is held fixed, the
incident wavevector ki is set by energy conservation,

$$k_i = \sqrt{k_f^2 + \frac{\Delta E}{\hbar^2/2m_n}},$$

so ki is not a free parameter: fixing kf and choosing dE determines it. The scan
runs from dE = 1 to 6 meV; in some plots I drop the 1 meV row, whose intensity is so
high that it saturates the colour scale and washes out the contrast along the rest of
the branch.

Although McStas derives ki internally from (kf, dE) via the relation above, I
cross-check the spectrometer geometry independently in TAKIN. The magnitude of the
total scattering vector follows from the scattering triangle,

$$|Q|^2 = k_i^2 + k_f^2 - 2 k_i k_f \cos(\text{stt}),$$

so sweeping stt at fixed kf changes both the magnitude and the direction of Q.

Because this is inelastic scattering, the relevant quantity is the reduced (phonon)
wavevector q, defined by

$$\mathbf{Q} = \mathbf{G} + \mathbf{q} \quad\Longleftrightarrow\quad
\mathbf{q} = \mathbf{Q} - \mathbf{G},$$

where G is a reciprocal-lattice vector (the Bragg point / zone centre nearest to Q)
and q lies in the first Brillouin zone. Changing Q therefore changes q, which is what
moves the measurement along the phonon dispersion. I believe the PANDA instrument is
measuring along  (0 0 2)→(0 0 4) zones, which matches at
dE = 2 meV and 3 meV.

```python
from IPython.display import clear_output



stt_vals  = np.arange(80, 160, 2.5)
kf_values=np.array([2.568, 0, 2.472, 0, 2.372, 0, 2.268, 0, 2.159, 0, 2.045])  # final wavevector values
dE_values = np.array([1,1.5,2,2.5,3,3.5,4,4.5,5,5.5,6], dtype=float)
I = np.full((len(dE_values), len(stt_vals)), np.nan)   # NaN = "not run yet"

fig, ax = plt.subplots(figsize=(7, 4))

for i, dE in enumerate(dE_values):
    instrument.set_parameters(kf=float(kf_values[i]))  # set kf for this dE row
    for j, stt in enumerate(stt_vals):
        instrument.set_parameters(dE=float(dE), stt=float(stt), sth=float(stt)/2,ki=2.66)
        data = instrument.backengine()
        PSD  = ms.name_search("PSD_det", data)
        I[i, j] = PSD.metadata.total_I

    # redraw after each completed dE row
    ax.clear()
    mesh = ax.pcolormesh(stt_vals, dE_values, I, shading="nearest")
    ax.set_xlabel("Sample two-theta  stt  (deg)")
    ax.set_ylabel("Energy transfer  dE  (meV)")
    ax.set_title(f"row {i+1}/{len(dE_values)}:  dE = {dE:.0f} meV done")
    if i == 0:
        cbar = fig.colorbar(mesh, ax=ax, label="Total intensity (PSD_det)")
    clear_output(wait=True)
    display(fig)

print(I)
```
